```python
#contain dublicate

#Examples:
nums1 = [1, 2, 3, 1]  #→ True  (1 appears twice)
nums2= [1, 2, 3, 4]  #→ False (all unique)
nums3 = [1, 1, 1, 3]  #→ True  (1 appears three times)

def contain_dublicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)

    return False

contain_dublicate(nums2)
```

```python
#two sum
nums = [3, 5, 8, 12]
target = 13

def two_sums(nums, target):
    seen = {}
    for i,num in enumerate(nums):
            complement = target - num
            if complement in seen:
                seen[num] = i
                return [seen[complement],i]
            
            seen[num] = i

    return []
    
two_sums(nums , target)
             
```

```python
#anagram

def anagram(atring : str, btring : str):
    freq = {}
    for s in atring:
        if s not in freq:
            freq[s] =  1
        else :
            freq[s] += 1

    for t in btring:
        if t in freq:
            freq[t] -= 1
        else:
            return False
    for t in freq.values():
        if t != 0:
            return False
    return True

anagram("sapplea", "ppsalea")
```

```python
#group anagram

def group_anagrams(strs):
    group = {}
    for s in strs:
       signature  = "".join(sorted(s))
       if signature not in group:
          group[signature] = [s]
       else:
          group[signature].append(s)

    return group.values()
          


```

```python
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

group_anagrams(strs)
```
